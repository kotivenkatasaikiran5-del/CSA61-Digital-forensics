```python


import hashlib
import shutil
import pandas as pd
from google.colab import files

print("Upload Sample Evidence Files")
uploaded = files.upload()

records = []

evidence_id = 1

for filename in uploaded.keys():

    # Create duplicate copy
    duplicate_file = "Duplicate_" + filename
    shutil.copy(filename, duplicate_file)

    # Create image copy (simulation)
    image_file = "Image_" + filename
    shutil.copy(filename, image_file)

    # Calculate SHA-256 Hash
    sha256 = hashlib.sha256()

    with open(filename, "rb") as f:

        while True:

            data = f.read(4096)

            if not data:
                break

            sha256.update(data)

    file_hash = sha256.hexdigest()

    records.append({
        "Evidence ID": "EV" + str(evidence_id).zfill(3),
        "Original File": filename,
        "Duplicate Copy": duplicate_file,
        "Image Copy": image_file,
        "SHA-256": file_hash
    })

    evidence_id += 1

df = pd.DataFrame(records)

print("\nImaging vs Duplication Report\n")
print(df)

output_file = "imaging_duplication_report.csv"

df.to_csv(output_file, index=False)

print("\nReport Saved Successfully")

files.download(output_file)

for file in df["Duplicate Copy"]:
    files.download(file)

for file in df["Image Copy"]:
    files.download(file)
```
