```python
 class FileSystemSimulator:

    def __init__(self):

        self.fs_data = {

            "FAT32": {
                "Maximum File Size": "4 GB",
                "Maximum Partition Size": "2 TB",
                "Security": "Basic (No Permissions)",
                "Journaling": "No",
                "Compression": "No",
                "Encryption": "No",
                "File Permissions": "No",
                "Performance": "Good",
                "Best Used For": "USB Drives, SD Cards"
            },

            "NTFS": {
                "Maximum File Size": "16 EB",
                "Maximum Partition Size": "256 TB",
                "Security": "Advanced",
                "Journaling": "Yes",
                "Compression": "Yes",
                "Encryption": "Yes",
                "File Permissions": "Yes",
                "Performance": "Very Good",
                "Best Used For": "Windows Operating System"
            },

            "EXT4": {
                "Maximum File Size": "16 TB",
                "Maximum Partition Size": "1 EB",
                "Security": "Unix Permissions",
                "Journaling": "Yes",
                "Compression": "No",
                "Encryption": "Supported",
                "File Permissions": "Yes",
                "Performance": "Excellent",
                "Best Used For": "Linux Operating System"
            }

        }

    def menu(self):

        print("\n" + "="*60)
        print("     FILE SYSTEM FEATURE SIMULATION")
        print("="*60)

        print("1. View All File System Features")
        print("2. Compare Two File Systems")
        print("3. Search File System")
        print("4. Exit")

    def display_all(self):

        print("\n")

        for fs, details in self.fs_data.items():

            print("="*60)
            print(fs)
            print("="*60)

            for key, value in details.items():
                print(f"{key:25}: {value}")

            print()

    def compare(self):

        fs1 = input("Enter First File System (FAT32/NTFS/EXT4): ").upper()
        fs2 = input("Enter Second File System (FAT32/NTFS/EXT4): ").upper()

        mapping = {
            "FAT32":"FAT32",
            "NTFS":"NTFS",
            "EXT4":"EXT4"
        }

        if fs1 not in mapping or fs2 not in mapping:
            print("Invalid File System")
            return

        fs1 = mapping[fs1]
        fs2 = mapping[fs2]

        print("\n")
        print("="*90)
        print(f"{'Feature':30}{fs1:25}{fs2}")
        print("="*90)

        for feature in self.fs_data[fs1]:

            print(f"{feature:30}{self.fs_data[fs1][feature]:25}{self.fs_data[fs2][feature]}")
    def search(self):

        fs = input("Enter File System Name: ").upper()

        if fs == "FAT32":
            fs = "FAT32"
        elif fs == "NTFS":
            fs = "NTFS"
        elif fs == "EXT4":
            fs = "EXT4"
        else:
            print("File System Not Found")
            return

        print("\n")
        print("="*60)
        print(fs)
        print("="*60)

        for key, value in self.fs_data[fs].items():
            print(f"{key:25}: {value}")

sim = FileSystemSimulator()

while True:

    sim.menu()

    choice = input("\nEnter Your Choice: ")

    if choice == "1":
        sim.display_all()

    elif choice == "2":
        sim.compare()

    elif choice == "3":
        sim.search()

    elif choice == "4":
        print("\nExiting Program...")
        print("Thank You!")
        break

    else:
        print("Invalid Choice!")
```
