```python
import hashlib
import pandas as pd
from google.colab import files

print("Upload Two Similar Files")
print("Example: original.txt and modified.txt")

uploaded = files.upload()

records = []

def calculate_sha256(filename):

    sha256 = hashlib.sha256()

    with open(filename, "rb") as file:

        while True:

            data = file.read(4096)

            if not data:
                break

            sha256.update(data)

    return sha256.hexdigest()


for filename in uploaded.keys():

    hash_value = calculate_sha256(filename)

    records.append({
        "File Name": filename,
        "SHA-256 Hash": hash_value
    })

df = pd.DataFrame(records)

print("\nHash Values\n")
print(df)

if len(df) == 2:

    if df.iloc[0]["SHA-256 Hash"] == df.iloc[1]["SHA-256 Hash"]:
        result = "No Avalanche Effect"
    else:
        result = "Avalanche Effect Observed"

    print("\nResult :", result)

    report = pd.DataFrame({
        "File 1":[df.iloc[0]["File Name"]],
        "File 2":[df.iloc[1]["File Name"]],
        "Hash 1":[df.iloc[0]["SHA-256 Hash"]],
        "Hash 2":[df.iloc[1]["SHA-256 Hash"]],
        "Observation":[result]
    })

    report.to_csv("avalanche_effect_report.csv",index=False)

    files.download("avalanche_effect_report.csv")

else:

    print("\nPlease upload exactly TWO files.")
```
