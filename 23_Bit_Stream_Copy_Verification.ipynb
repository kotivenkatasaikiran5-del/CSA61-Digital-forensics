```python
 import hashlib
import shutil
import pandas as pd
from google.colab import files

print("Upload Evidence Files")
uploaded = files.upload()

records = []

evidence_id = 1

def calculate_hash(file_name):

    sha256 = hashlib.sha256()

    with open(file_name, "rb") as file:

        while True:

            data = file.read(4096)

            if not data:
                break

            sha256.update(data)

    return sha256.hexdigest()

for filename in uploaded.keys():

    # Create Bit Stream Copy
    copy_name = "BitStream_" + filename
    shutil.copyfile(filename, copy_name)

    original_hash = calculate_hash(filename)
    copy_hash = calculate_hash(copy_name)

    if original_hash == copy_hash:
        status = "Verified"
    else:
        status = "Verification Failed"

    records.append({
        "Evidence ID": "EV" + str(evidence_id).zfill(3),
        "Original File": filename,
        "Bit Stream Copy": copy_name,
        "Original SHA-256": original_hash,
        "Copy SHA-256": copy_hash,
        "Status": status
    })

    evidence_id += 1

df = pd.DataFrame(records)

print("\nBit Stream Copy Verification Report\n")
print(df)

output_file = "bit_stream_verification_report.csv"

df.to_csv(output_file, index=False)

print("\nReport Saved Successfully")

files.download(output_file)

for file in df["Bit Stream Copy"]:
    files.download(file)
```
