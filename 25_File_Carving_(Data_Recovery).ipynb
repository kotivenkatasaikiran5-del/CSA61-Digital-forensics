```python
 import os
from google.colab import files

print("=" * 60)
print("        FILE CARVING (DATA RECOVERY)")
print("=" * 60)

print("\nUpload a binary image or raw file:")
uploaded = files.upload()

filename = list(uploaded.keys())[0]

# Output Folder
output_folder = "Recovered_Files"
os.makedirs(output_folder, exist_ok=True)

with open(filename, "rb") as f:
    data = f.read()

signatures = {
    "jpg": {
        "start": b"\xff\xd8",
        "end": b"\xff\xd9"
    },
    "png": {
        "start": b"\x89PNG\r\n\x1a\n",
        "end": b"IEND\xaeB`\x82"
    },
    "pdf": {
        "start": b"%PDF",
        "end": b"%%EOF"
    },
    "zip": {
        "start": b"PK\x03\x04",
        "end": b"PK\x05\x06"
    }
}

total = 0

for extension, sig in signatures.items():

    start = 0

    while True:

        start_index = data.find(sig["start"], start)

        if start_index == -1:
            break

        end_index = data.find(sig["end"], start_index)

        if end_index == -1:
            break

        end_index += len(sig["end"])

        carved = data[start_index:end_index]

        outfile = os.path.join(
            output_folder,
            f"Recovered_{total}.{extension}"
        )

        with open(outfile, "wb") as out:
            out.write(carved)

        print(f"[+] Recovered: {outfile}")

        total += 1
        start = end_index

print("\n------------------------------------")
print("Recovery Completed")
print("------------------------------------")
print("Total Files Recovered:", total)

if total == 0:
    print("No supported files found.")
else:
    print("\nRecovered files:")

    for f in os.listdir(output_folder):
        print(" -", f)

print("\nDownload recovered files manually from the folder if needed.")
```
